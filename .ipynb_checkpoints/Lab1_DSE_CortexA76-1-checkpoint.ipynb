````python
import numpy as np, pandas as pd, json, re, sys, platform
import matplotlib.pyplot as plt
import matplotlib as mpl
mpl.rcParams.update({"figure.dpi": 110, "axes.spines.top": False, "axes.spines.right": False,
                     "axes.grid": True, "grid.alpha": .25, "font.size": 10, "legend.frameon": False})
# Paleta apta para daltonismo (Okabe-Ito)
AZUL, NARANJA, VERDE, ROJO, GRIS, NEGRO = "#0072B2", "#E69F00", "#009E73", "#D55E00", "#8c8c8c", "#222222"

d = pd.read_csv(DATA / "results.csv")
rl = pd.read_csv(DATA / "random_log.csv")
man = d[d.source == "manual"].copy()
nsga = d[d.source == "nsga"].sort_values("eval_idx").reset_index(drop=True)
nm = d[d.source != "manual"].drop_duplicates("hash").set_index("hash")      # NSGA + aleatorias, sin manuales
rnd = nm.loc[rl.hash].reset_index()                                         # las 300 evaluaciones aleatorias EN ORDEN
BASE = man[man.name == "sim1"].iloc[0]                                      # A76 por defecto
KEYS = ["iq_entries", "width", "rob_entries", "lq_entries", "sq_entries", "l1i_kb", "l1d_kb", "l2_kb", "issue_width"]
print(d.source.value_counts().to_dict(), "| evaluaciones aleatorias:", len(rnd),
      "(", rl.hash.isin(set(nsga.hash)).sum(), "ya estaban simuladas por el NSGA-II )")

def pareto_mask(P):
    # True para los puntos no dominados (minimizar ambas columnas)
    P = np.asarray(P); m = np.ones(len(P), bool)
    for i, (a, b) in enumerate(P):
        m[i] = not ((P[:, 0] <= a) & (P[:, 1] <= b) & ((P[:, 0] < a) | (P[:, 1] < b))).any()
    return m

def hypervolume(P, ref):
    # Hipervolumen 2D (área dominada respecto a ref) para minimizar ambos objetivos
    P = np.asarray(P); P = P[np.lexsort((P[:, 1], P[:, 0]))]
    h, best = 0.0, ref[1]
    for c, e in P:
        if c < ref[0] and e < best:
            h += (ref[0] - c) * (best - e); best = e
    return h
````

## 1. Resumen

Se exploró el espacio de diseño de un Cortex-A76 (gem5 v25.0.0.1 en modo SE + McPAT) con el
workload `mp3_enc`, minimizando simultáneamente **CPI** y **energía** (la métrica de la guía:
Energía = (Total Leakage + Runtime Dynamic) × CPI; EDP = Energía × CPI).

El camino fue: (1) perfilar la aplicación, (2) 40 simulaciones manuales para entender qué parámetros
importan y cuáles ve McPAT, (3) fijar un espacio de 23 328 combinaciones, (4) un DSE con **NSGA-II
de 1004 simulaciones únicas**, y (5) una búsqueda **aleatoria de 300 evaluaciones** como línea base.

````python
best_cpi, best_en, best_edp = [nsga.loc[nsga[c].idxmin()] for c in ("cpi", "energy", "edp")]
def pct(a, b): return 100 * (a - b) / b
resumen = pd.DataFrame({
    "Config. por defecto (sim1)": [BASE.cpi, BASE.energy, BASE.edp],
    "Mejor CPI (NSGA-II)":       [best_cpi.cpi, best_cpi.energy, best_cpi.edp],
    "Mejor energía (NSGA-II)":   [best_en.cpi, best_en.energy, best_en.edp],
    "Mejor EDP (NSGA-II)":       [best_edp.cpi, best_edp.energy, best_edp.edp]}, index=["CPI", "Energía", "EDP"])
display(resumen.round(5))
print("Mejor CPI   frente a la base: CPI %+.2f %%" % pct(best_cpi.cpi, BASE.cpi))
print("Mejor energía frente a la base: energía %+.2f %%" % pct(best_en.energy, BASE.energy))
print("Mejor EDP   frente a la base: EDP %+.2f %%  (CPI %+.2f %%, energía %+.2f %%)" % (
      pct(best_edp.edp, BASE.edp), pct(best_edp.cpi, BASE.cpi), pct(best_edp.energy, BASE.energy)))
````

**Conclusiones principales** (las cifras exactas salen de las celdas de arriba y de las secciones 9 y 10):

1. La mejor configuración por EDP reduce el EDP en ≈ 26 % frente al A76 por defecto (CPI ≈ 18 % menor, energía ≈ 9.5 % menor). La de menor CPI lo baja 19.2 % y la de menor energía 10.6 %.
2. Casi toda la mejora frente al A76 por defecto viene de subir la cola de instrucciones (**IQ**, de 16 a 48–64). Dentro del espacio del DSE, el **ancho** (4 frente a 3) es lo que más mueve el CPI
   y el **ROB** lo que más mueve la energía; las cachés son ajustes finos (≈ 1–2 % en CPI) porque `mp3_enc` tiene un conjunto de trabajo pequeño.
3. El NSGA-II supera a la búsqueda aleatoria con el mismo presupuesto, pero por **poco en el mejor punto** y por **mucho en la calidad del frente de Pareto**
   (sección 9.5). El espacio ya venía muy acotado por los experimentos manuales.
4. Limitación: varias soluciones del frente están en el borde inferior del espacio (SQ = 32, L2 = 128 kB); el óptimo de energía podría estar más allá.

| Punto flotante | ≈ 9.2 % (FloatAdd 3.52, FloatMult 3.05, FloatCmp 0.93…); SIMD 0.66 % | FP **escalar**, casi nada vectorial |

| FU ocupada (`fuBusyRate`) | 3.02 % de las instrucciones; 84.4 % son puertos de memoria | Poco limitante |
| Predictor de saltos | 6.21 % de instrucciones consultan; 16.95 % de squashes | 4.58 % de lo emitido se descarta |

**Qué significa.** La jerarquía de memoria no limita (el conjunto de trabajo cabe en L1); el cuello de botella está en el **núcleo fuera de orden**
(la ventana de instrucciones y el predictor). Decisión: probar primero parámetros del núcleo (IQ, ROB, colas, anchos) y tratar las cachés como un eje de **costo** más que de rendimiento.

````python
# (Opcional) Recalcular el perfil desde un stats.txt de gem5, si se tiene a mano.
def perfil_desde_stats(path):
    t = Path(path).read_text(errors="replace")
    def g(pat):
        m = re.search(pat, t, re.M); return float(m.group(1)) if m else float("nan")
    tot = g(r"^simInsts\s+(\d+)")
    out = {"instrucciones": tot}
    for k in ("MemRead", "MemWrite", "IntAlu"):
        out["% " + k] = 100 * g(r"^system\.cpu\.commit\.committedInstType_0::%s\s+(\d+)" % k) / tot
    out["L1D miss %"] = 100 * g(r"^system\.cpu\.dcache\.overallMissRate::total\s+(\S+)")
    return out
# Ejemplo:  perfil_desde_stats("stats.txt")
````

## 4. Experimentos manuales y qué ve McPAT

Antes de definir el DSE se hicieron **40 simulaciones manuales** (base, variantes de un parámetro, combinaciones alrededor de IQ = 64, cachés, anchos, `issue_width`).
Sirvieron para tres cosas: saber qué parámetros mueven el CPI, saber cuáles **ve McPAT**, y elegir rangos razonables para el DSE.

````python
IQ64 = man[man.name == "iq64"].iloc[0]
t = man[["name", "iq_entries", "width", "rob_entries", "l1i_kb", "l1d_kb", "l2_kb", "issue_width", "cpi", "power_w", "energy", "edp", "area_mm2"]].copy()
t["ΔEDP vs sim1 %"] = 100 * (t.edp / BASE.edp - 1)
t["ΔEDP vs iq64 %"] = 100 * (t.edp / IQ64.edp - 1)
pd.set_option("display.width", 220, "display.max_rows", 100)
display(t.round(4).reset_index(drop=True))
````

### 4.1 Qué parámetros llegan a McPAT

Se comprobó con `diff config.xml` entre una variante y la base, y con el cambio de área/potencia de cada una.

| Parámetro de gem5 | ¿Lo ve McPAT? | Consecuencia |
|---|---|---|
| `rob_entries`, `lq_entries`, `sq_entries` | Sí | Se cobran en potencia y área |
| `iq_entries` | Sí, **pero la mitad** (`instruction_window_size`) | Desplaza la curva de potencia |
| anchos de fetch/decode/commit, `issue_width`, tamaños de caché | Sí | Se cobran |
| `num_fu_*` (puertos), predictor de saltos | **No** | Serían "gratis" artificiales: se dejan fijos |
| `dispatch_width`, `wb_width` | **No** (y no cambian el CPI) | Se dejan fijos |

Ejemplo del riesgo: el predictor TAGE mejora el CPI 1.86 % con potencia idéntica, porque McPAT no modela el predictor. Un optimizador lo "elegiría" siempre, pero es un artefacto del modelo, no una mejora real.

### 4.2 Hallazgos de los experimentos manuales

1. **IQ domina.** Con IQ = 16 (defecto) casi nada más mueve el CPI. IQ 64 baja el CPI 19 % y el EDP 23.7 %; satura entre 48 y 64 (128 no mejora y cuesta potencia). IQ 8 es desastroso (EDP +56.7 %).
2. **Interacciones.** El efecto de cada parámetro depende de los demás: el ancho 2 cuesta +4.45 % de CPI con IQ = 16 y +15 % con IQ = 64; ROB 64 no cuesta nada con IQ = 16 y +1.9 % con IQ = 64.
   Por eso no sirve optimizar de a un parámetro: es el argumento para una búsqueda multiobjetivo.
3. **Ancho 4 es el punto correcto**: ancho 3 cuesta +13 % de CPI; ancho 5 y 6 no mejoran (se probó también subir `wb_width`: ruido).
4. **`issue_width`** (defecto 8) sobra en un pipeline de ancho 4: bajar a 4 casi no cambia el CPI (+0.04 %) y McPAT cobra menos potencia y área. Es una variable legítima del DSE porque McPAT sí la ve.
5. **Cachés: poco efecto en CPI** (< 0.5 %) y mucho en costo. L2 de 2 MB cuesta +7.8 % de área por −0.25 % de CPI; por eso L2 ≥ 2 MB se descartó.

## 5. Rendimiento frente a recursos

### 5.1 Barridos de un parámetro (simulaciones manuales)
Cada curva cambia **un** parámetro; el resto queda en la base. Para IQ la base es el A76 por defecto; para los demás es la configuración con IQ = 64
(donde vive el DSE). Se muestran CPI, potencia y EDP (cada métrica en su propio eje, sin ejes dobles).

````python
def sweep(param, base, label_vals=None):
    # filas manuales sin extras donde TODOS los demás parámetros del espacio valen lo de 'base'
    others = [k for k in KEYS if k != param]
    ok = man[man.extras.isna()]
    for k in others: ok = ok[ok[k] == base[k]]
    return ok.sort_values(param)

IQ16 = {k: BASE[k] for k in KEYS}
IQ64d = {k: IQ64[k] for k in KEYS}
panels = [("iq_entries", IQ16, "Entradas de la IQ"), ("width", IQ64d, "Ancho (fetch/decode/rename/commit)"),
          ("issue_width", IQ64d, "issue_width"), ("rob_entries", IQ64d, "Entradas del ROB"),
          ("l1i_kb", IQ64d, "L1I (kB)"), ("l1d_kb", IQ64d, "L1D (kB)"), ("l2_kb", IQ64d, "L2 (kB)")]
for met, nombre, col in (("cpi", "CPI", AZUL), ("power_w", "Potencia (W)", NARANJA), ("edp", "EDP", VERDE)):
    fig, axs = plt.subplots(2, 4, figsize=(13, 5.2)); axs = axs.ravel()
    for ax, (p, base, xl) in zip(axs, panels):
        s = sweep(p, base)
        ax.plot(s[p].astype(str), s[met], "o-", color=col, lw=2, ms=6)
        ax.set_xlabel(xl); ax.set_ylabel(nombre)
        ref = base
        ax.axvline(str(int(base[p])) if str(int(base[p])) in list(s[p].astype(str)) else 0, color=GRIS, ls=":", lw=1)
    axs[-1].axis("off")
    fig.suptitle(f"{nombre} frente a cada recurso (un parámetro a la vez)", y=1.0)
    plt.tight_layout(); plt.savefig(FIG / f"fig_barrido_{met}.png", dpi=150); plt.show()
````

**Cómo leerlas.** La línea punteada marca el valor de la base. El CPI cae mucho al subir la IQ hasta ~ 48–64 y luego se aplana (rendimientos decrecientes) mientras la potencia sigue subiendo,
así que el EDP tiene un mínimo cerca de IQ 48–64. Ancho 3 empeora el CPI de golpe; pasar de 4 a 5 o 6 solo suma potencia. `issue_width` casi no mueve el CPI entre 3 y 8 pero sí la potencia.
Las cachés prácticamente no cambian el CPI y sí la potencia (L2 sobre todo).

### 5.2 Efecto medio de cada parámetro sobre una muestra **sin sesgo**
Las 300 configuraciones aleatorias son una muestra uniforme del espacio, así que su promedio por valor de cada parámetro estima el efecto *marginal* sin el sesgo de una búsqueda
(en el NSGA-II el promedio por valor está sesgado porque el optimizador concentra las muestras en la zona buena).

````python
labels = {"iq_entries": "IQ", "width": "Ancho", "rob_entries": "ROB", "lq_entries": "LQ", "sq_entries": "SQ",
          "l1i_kb": "L1I (kB)", "l1d_kb": "L1D (kB)", "l2_kb": "L2 (kB)", "issue_width": "issue_width"}
for met, nombre, col in (("cpi", "CPI medio", AZUL), ("energy", "Energía media", NARANJA)):
    fig, axs = plt.subplots(3, 3, figsize=(11, 8))
    for ax, k in zip(axs.ravel(), KEYS):
        g = rnd.groupby(k)[met].agg(["mean", "std", "count"])
        ax.errorbar(g.index.astype(str), g["mean"], yerr=g["std"] / np.sqrt(g["count"]), fmt="o-", color=col, capsize=3, lw=2)
        ax.set_xlabel(labels[k]); ax.set_ylabel(nombre)
    fig.suptitle(f"{nombre} por valor de cada parámetro (300 configuraciones aleatorias; barras = error estándar)", y=1.0)
    plt.tight_layout(); plt.savefig(FIG / f"fig_marginal_{met}.png", dpi=150); plt.show()

# Tabla de importancia: cuánto cambia el CPI medio entre el mejor y el peor valor de cada parámetro
imp = pd.DataFrame({labels[k]: {"rango del CPI medio": rnd.groupby(k).cpi.mean().max() - rnd.groupby(k).cpi.mean().min(),
                                "rango de la energía media": rnd.groupby(k).energy.mean().max() - rnd.groupby(k).energy.mean().min()}
                    for k in KEYS}).T.sort_values("rango del CPI medio", ascending=False)
display(imp.round(4))
````

**Lectura.** Dentro de los rangos del DSE el **ancho** es, con diferencia, lo que más mueve el CPI (≈ 0.10 entre 3 y 4); el resto de parámetros mueve el CPI ≈ 0.003–0.015 (menos del 2 %).
La IQ pesa muy poco *aquí* porque el espacio empieza en 32, donde ya está en la zona de saturación; su efecto grande ocurre entre 8 y 32 (sección 5.1). En energía manda el **ROB** (≈ 0.09), seguido del ancho (≈ 0.05):
son los parámetros que McPAT cobra más. Esto confirma, con una muestra uniforme, lo que mostraron los experimentos manuales.

Notas: el **ancho** es `fetch = decode = rename = commit` a la vez; `issue_width` baja de 8 a 3–6 porque el pipeline es de ancho 4 y McPAT sí cobra ese costo.

**Fijos** (y por qué): puertos de unidades funcionales y predictor de saltos (McPAT no los ve, serían "gratis" artificiales), `dispatch_width` y `wb_width` (no cambian el CPI ni los ve McPAT),
L3 (2 MB), asociatividades y latencias. Probar todo el espacio tomaría ≈ 33 días de cómputo; 1000 evaluaciones son el 4.3 % del espacio.

### 6.2 Objetivos
Minimizar **CPI** y **energía** (Energía = (Leakage + Dynamic) × CPI). El **EDP** no es un objetivo: se calcula después. Con CPI casi constante, EDP y energía serían casi la misma variable
y el algoritmo optimizaría dos veces lo mismo.

---
## 7. Estrategias de búsqueda

**Fuerza bruta descartada**: 23 328 simulaciones ≈ 33 días. **Se usó una estrategia más inteligente que la fuerza bruta** (lo que la guía premia):

1. **Reducción manual del espacio** con las 40 simulaciones de las secciones 4 y 5 (de rangos enormes a 23 328 combinaciones razonables).
2. **NSGA-II** (algoritmo genético multiobjetivo, `pymoo` 0.6) sobre CPI y energía:
   - Variables enteras codificadas como índice de cada lista de valores.
   - Población 50, semilla 42; cruce SBX (prob. 0.9, η = 3); mutación polinómica (prob. 0.2, η = 3); `RoundingRepair`; sin duplicados.
   - Población inicial: 6 semillas de los experimentos manuales (IQ 64 con issue 3, 4 y 6; ROB 64; L1D 32 kB; IQ 48) + 44 aleatorias.
   - Parada: 1000 simulaciones **únicas** nuevas (terminó con 1004 en 29 generaciones).
3. **Búsqueda aleatoria de 300 evaluaciones** (semilla 123) como línea base, para medir cuánto aporta de verdad el algoritmo evolutivo.
4. **Caché** por hash SHA-1 de los parámetros: nunca se simula dos veces la misma configuración y el DSE es reanudable (volver a lanzar el mismo comando).
5. **Paralelismo**: 10 simulaciones a la vez (`ProcessPoolExecutor`), un único proceso escribe el CSV.

---
## 8. Automatización (código)

Todo el flujo está en dos scripts que este notebook **escribe a disco**: `scripts/sim.sh` (una simulación a mano, usada en las fases 0–2) y `dse.py` (motor paralelo, caché, NSGA-II, aleatoria).
Con `RUN_SIMS = False` (valor por defecto) solo se escriben los archivos y se analizan los resultados ya guardados; con `RUN_SIMS = True` se lanzan simulaciones reales
(requiere gem5 y McPAT instalados en las rutas de `dse.py`).

### 8.1 Una simulación a mano: la función `sim`

````python
%%writefile scripts/sim.sh
# Uso:  source scripts/sim.sh ; sim nombre_carpeta --iq_entries=64 --issue_width=4
# Una simulación = gem5 -> conversor del profesor -> McPAT. SIEMPRE rutas absolutas y --cwd.
sim() { local d=$HOME/gem5/pruebas_lab1/$1; shift; mkdir -p $d
  ( cd $HOME/gem5 && /usr/bin/time -v ./build/ARM/gem5.fast -d $d \
    scripts/CortexA76_scripts_gem5/CortexA76.py -c $HOME/gem5/workloads/mp3_enc/mp3_enc \
    -o "$HOME/gem5/workloads/mp3_enc/mp3enc_testfile.wav mp3enc_outfile.mp3" \
    --cwd=$d "$@" > $d/salida.log 2> $d/time.log )
  ( cd $d && python3 ~/gem5/scripts/McPAT/gem5toMcPAT_cortexA76.py stats.txt config.json ~/gem5/scripts/McPAT/ARM_A76_2.1GHz.xml \
      && ~/mcpat/mcpat -infile config.xml -print_level 1 > potencia.txt ); }
# Tandas de 6 simulaciones en paralelo (fases 1-2):
#   export -f sim; printf '%s\n' "iq64 --iq_entries=64" ... | xargs -P 6 -L 1 bash -c 'sim "$@"' _
````

### 8.2 El motor del DSE: `dse.py`
Contiene: lectura del espacio desde `config/` (sección 6.1), invocación de gem5, conversión + McPAT, caché con `results.csv`, NSGA-II, búsqueda aleatoria, importación de las simulaciones manuales, `check` (comparar entornos) y `status`.

````python
%%writefile dse.py
#!/usr/bin/env python3
"""dse.py - Motor de simulación paralelo + NSGA-II + búsqueda aleatoria (Lab 1, Arq. Avanzadas).

Uso (desde cualquier carpeta):
  python3 dse.py import               # carga las 41 simulaciones manuales al results.csv
  python3 dse.py check                # 6 simulaciones reales; compara 3 con lo medido a mano
  python3 dse.py nsga --target 1000   # NSGA-II hasta 1000 simulaciones únicas nuevas
  python3 dse.py random --n 300       # línea base aleatoria
  python3 dse.py status               # resumen de results.csv
Opciones globales: --workers 6, --mock (simulador falso, para probar sin gem5).
El espacio de diseño y las semillas NO están en el código: se leen de config/espacio.csv y config/semillas.csv
(otra carpeta con DSE_CONFIG=/ruta). Cambiar rangos = editar el CSV; los resultados ya guardados se reconocen por el hash.
Reanudar = volver a lanzar el mismo comando: lo ya simulado sale de la caché (results.csv).
"""
import argparse, csv, hashlib, json, os, random, re, shutil, subprocess, sys, time
from concurrent.futures import ProcessPoolExecutor, as_completed
from datetime import datetime
from pathlib import Path

# ============================ RUTAS ============================
HOME   = Path.home()
GEM5   = HOME / "gem5"
MCPAT  = HOME / "mcpat" / "mcpat"
CONV   = GEM5 / "scripts/McPAT/gem5toMcPAT_cortexA76.py"
XML    = GEM5 / "scripts/McPAT/ARM_A76_2.1GHz.xml"
SCRIPT = "scripts/CortexA76_scripts_gem5/CortexA76.py"
BIN    = GEM5 / "workloads/mp3_enc/mp3_enc"
WAV    = GEM5 / "workloads/mp3_enc/mp3enc_testfile.wav"
REPO   = Path(__file__).resolve().parent
RES    = REPO / "results"
SIMS   = GEM5 / "pruebas_lab1" / "dse"

# ============ ESPACIO DE DISEÑO (se lee de config/*.csv, no está en el código) ============
# config/espacio.csv : parametro,defecto,valores,comentario   (valores separados por ';')
#                      El orden de las filas es el orden de las variables del optimizador.
# config/semillas.csv: una fila por semilla; columnas = parámetros (celda vacía = valor por defecto).
# Para otra carpeta de configuración:  DSE_CONFIG=/ruta/a/otra_config python3 dse.py ...
CONFIG = Path(os.environ.get("DSE_CONFIG", REPO / "config"))

def load_space(path=None):
    path = Path(path or CONFIG / "espacio.csv")
    if not path.exists():
        sys.exit(f"No encuentro {path}. Copia la carpeta config/ junto a dse.py (o usa DSE_CONFIG).")
    space, default = {}, {}
    with open(path, newline="", encoding="utf-8") as f:
        for r in csv.DictReader(f):
            space[r["parametro"].strip()] = [int(v) for v in r["valores"].split(";")]
            default[r["parametro"].strip()] = int(r["defecto"])
    return space, default

def load_seeds(path=None):
    path = Path(path or CONFIG / "semillas.csv")
    if not path.exists():
        return []
    with open(path, newline="", encoding="utf-8") as f:
        return [{k: int(row[k]) if row.get(k, "").strip() else DEFAULT[k] for k in KEYS} for row in csv.DictReader(f)]

SPACE, DEFAULT = load_space()   # SPACE: variable -> valores permitidos (el optimizador usa el índice)
KEYS = list(SPACE)              # DEFAULT: valores del A76 por defecto
SEEDS = load_seeds()            # puntos con los que arranca la población del NSGA-II

def full(p):            return {**DEFAULT, **p}
def phash(p):           return hashlib.sha1(json.dumps(full(p), sort_keys=True).encode()).hexdigest()[:10]
def decode(x):          return {k: SPACE[k][int(round(v))] for k, v in zip(KEYS, x)}
def encode(p):          return [SPACE[k].index(p[k]) for k in KEYS]

def cli_opts(p):
    """Solo pasa a gem5 lo que difiere del valor por defecto (igual que las simulaciones manuales)."""
    o = []
    for k, v in full(p).items():
        if k.startswith("x_"):                 # parámetros extra de las simulaciones manuales
            o.append(f"--{k[2:]}={v}")
        elif v == DEFAULT[k]:
            continue
        elif k == "width":
            o += [f"--{n}_width={v}" for n in ("fetch", "decode", "rename", "commit")]
        elif k.endswith("_kb"):
            o.append(f"--{k[:-3]}_size={v}kB")  # l1i_kb -> --l1i_size=32kB
        else:
            o.append(f"--{k}={v}")
    return o

# ========================= UNA SIMULACIÓN =========================
def grab(pat, txt):
    m = re.search(pat, txt, re.M)
    return float(m.group(1)) if m else None

def mock_sim(p):
    q = full(p)
    cpi = 0.80 + 6.0 / q["iq_entries"] + (0.12 if q["width"] == 3 else 0) + \
          {3: 0.004, 4: 0.0005, 6: 0.0002, 8: 0}.get(q["issue_width"], 0) + 0.0004 * 64 / q["l1d_kb"]
    pw = 1.2 + 0.012 * q["iq_entries"] + 0.15 * q["width"] + 0.003 * q["rob_entries"] + \
         0.002 * (q["lq_entries"] + q["sq_entries"]) + 0.0005 * (q["l1i_kb"] + q["l1d_kb"]) + \
         0.0004 * q["l2_kb"] + 0.03 * q["issue_width"]
    return {"cpi": cpi, "leak": 0.3 * pw, "dyn": 0.7 * pw, "area": 30 + 0.01 * q["l2_kb"],
            "host": 1.0, "gem5_wall": 1.0, "mcpat": 0.1}

def run_sim_once(p, root, mock):
    if mock:
        time.sleep(0.01)
        return mock_sim(p)
    d = Path(root) / phash(p)
    shutil.rmtree(d, ignore_errors=True)
    d.mkdir(parents=True)
    t0 = time.time()
    cmd = ["./build/ARM/gem5.fast", "-d", str(d), SCRIPT, "-c", str(BIN),
           "-o", f"{WAV} mp3enc_outfile.mp3", f"--cwd={d}"] + cli_opts(p)   # misma invocación que `sim`
    with open(d / "salida.log", "w") as f:
        rc = subprocess.run(cmd, cwd=GEM5, stdout=f, stderr=subprocess.STDOUT).returncode
    t_gem5 = time.time() - t0
    if rc != 0 or not (d / "stats.txt").exists():
        return {"error": f"gem5 salió con código {rc} (ver {d}/salida.log)"}
    t1 = time.time()
    r = subprocess.run(["python3", str(CONV), "stats.txt", "config.json", str(XML)],
                       cwd=d, capture_output=True)
    if r.returncode != 0:
        return {"error": f"conversor falló: {r.stderr.decode(errors='replace')[-200:]}"}
    with open(d / "potencia.txt", "w") as f:
        r = subprocess.run([str(MCPAT), "-infile", "config.xml", "-print_level", "1"],
                           cwd=d, stdout=f, stderr=subprocess.STDOUT)
    t_mcpat = time.time() - t1                      # conversión + McPAT
    st = (d / "stats.txt").read_text(errors="replace")
    pw = (d / "potencia.txt").read_text(errors="replace")
    res = {"cpi": grab(r"^system\.cpu\.cpi\s+(\S+)", st), "host": grab(r"^hostSeconds\s+(\S+)", st),
           "leak": grab(r"Total Leakage = ([\d.]+)", pw), "dyn": grab(r"Runtime Dynamic = ([\d.]+)", pw),
           "area": grab(r"Area = ([\d.]+)", pw), "gem5_wall": t_gem5, "mcpat": t_mcpat}
    if any(res[k] is None for k in ("cpi", "leak", "dyn", "area")):
        return {"error": f"no pude leer las métricas en {d}"}
    (d / "params.json").write_text(json.dumps(full(p)))
    for x in d.iterdir():                           # conservar solo lo necesario
        if x.name not in ("stats.txt", "potencia.txt", "params.json"):
            shutil.rmtree(x) if x.is_dir() else x.unlink()
    return res

def run_sim(p, root, mock):
    for intento in (1, 2):                          # un reintento por si fue un fallo puntual
        try:
            res = run_sim_once(p, root, mock)
        except Exception as e:                      # noqa
            res = {"error": repr(e)}
        if "error" not in res:
            return res
    return res

# ===================== RESULTADOS (CACHÉ) =====================
COLS = ["hash", "source", "name", "gen", "eval_idx"] + KEYS + \
       ["extras", "cpi", "ipc", "leakage_w", "dynamic_w", "power_w", "energy", "edp", "area_mm2",
        "gem5_host_s", "gem5_wall_s", "mcpat_s", "timestamp"]

class Store:
    def __init__(self, path):
        self.path, self.rows = Path(path), {}
        if self.path.exists():
            with open(self.path) as f:
                for r in csv.DictReader(f):
                    self.rows[self.key(r)] = r
    @staticmethod
    def key(r):
        # Las filas manuales (otra máquina/entorno) nunca sirven de caché para el DSE
        return ("m:" if r["source"] == "manual" else "") + r["hash"]
    def add(self, row):
        self.path.parent.mkdir(parents=True, exist_ok=True)
        nuevo = not self.path.exists()
        with open(self.path, "a", newline="") as f:
            w = csv.DictWriter(f, COLS)
            if nuevo:
                w.writeheader()
            w.writerow(row)
            f.flush(); os.fsync(f.fileno())
        self.rows[self.key(row)] = row
    def count(self, source):
        return sum(1 for r in self.rows.values() if r["source"] == source)
    def best(self, col):
        v = [r for r in self.rows.values() if r.get(col) not in (None, "")]
        return min(v, key=lambda r: float(r[col])) if v else None

def make_row(p, source, name, gen, idx, cpi, power, energy, edp, area, leak="", dyn="", host="", wall="", mc=""):
    q = full(p)
    extras = {k: v for k, v in q.items() if k.startswith("x_")}
    row = {"hash": phash(p), "source": source, "name": name, "gen": gen, "eval_idx": idx,
           "extras": json.dumps(extras) if extras else "", "cpi": cpi, "ipc": 1 / cpi,
           "leakage_w": leak, "dynamic_w": dyn, "power_w": power, "energy": energy, "edp": edp,
           "area_mm2": area, "gem5_host_s": host, "gem5_wall_s": wall, "mcpat_s": mc,
           "timestamp": datetime.now().isoformat(timespec="seconds")}
    row.update({k: q[k] for k in KEYS})
    return row

def evaluate_batch(store, plist, source, gen, args):
    """Devuelve [(cpi, energía)] en el mismo orden de plist. Simula solo lo que no está en la caché."""
    todo = {}
    for p in plist:
        h = phash(p)
        if h not in store.rows and h not in todo:
            todo[h] = p
    if todo:
        with ProcessPoolExecutor(args.workers) as ex:
            futs = {ex.submit(run_sim, p, args.root, args.mock): p for p in todo.values()}
            for i, fu in enumerate(as_completed(futs), 1):
                p = futs[fu]
                try:
                    res = fu.result()
                except Exception as e:             # noqa
                    res = {"error": repr(e)}
                if "error" in res:
                    print(f"  ! FALLÓ {phash(p)} {p}: {res['error']}", flush=True)
                    continue
                pw = res["leak"] + res["dyn"]
                en = pw * res["cpi"]
                store.add(make_row(p, source, "", gen, len(store.rows) + 1, res["cpi"], pw, en,
                                   en * res["cpi"], res["area"], res["leak"], res["dyn"],
                                   res["host"], round(res["gem5_wall"], 1), round(res["mcpat"], 1)))
                print(f"  [{i}/{len(todo)}] {phash(p)} cpi={res['cpi']:.6f} pot={pw:.4f} "
                      f"gem5={res['gem5_wall']:.0f}s mcpat={res['mcpat']:.0f}s", flush=True)
    out = []
    for p in plist:
        r = store.rows.get(phash(p))
        out.append((float(r["cpi"]), float(r["energy"])) if r else (1e3, 1e3))   # penalización si falló
    return out

# ====================== SIMULACIONES MANUALES ======================
# (nombre, cambios respecto del A76 por defecto, CPI, potencia W, energía, EDP, área mm2)
# Claves x_*: parámetros fuera del espacio del DSE (puertos, predictor, wb/dispatch).
MANUAL = [
    ('sim1', {}, 1.045577, 2.168, 2.26681, 2.37013, 35.5383),
    ('ancho2', {'width': 2}, 1.092134, 2.0775, 2.26893, 2.47797, 35.5238),
    ('fur4', {'x_num_fu_read': 4}, 1.043534, 2.1665, 2.26077, 2.35919, 35.5383),
    ('fuw2', {'x_num_fu_write': 2}, 1.047313, 2.1656, 2.26804, 2.37535, 35.5383),
    ('lq96', {'lq_entries': 96}, 1.045577, 2.172, 2.27096, 2.37446, 35.5436),
    ('rob256', {'rob_entries': 256}, 1.045533, 2.2494, 2.35177, 2.45886, 36.521),
    ('sq96', {'sq_entries': 96}, 1.045595, 2.1742, 2.27335, 2.377, 35.5428),
    ('bp_tage', {'x_branch_predictor_type': 8}, 1.026128, 2.1681, 2.22479, 2.28292, 35.5383),
    ('iq32', {'iq_entries': 32}, 0.864596, 2.46, 2.12687, 1.83888, 35.5389),
    ('iq64', {'iq_entries': 64}, 0.847006, 2.5208, 2.13517, 1.80851, 35.5415),
    ('lq32', {'lq_entries': 32}, 1.045585, 2.162, 2.26055, 2.3636, 35.5306),
    ('rob64', {'rob_entries': 64}, 1.045596, 2.1363, 2.23373, 2.33558, 35.1608),
    ('sq32', {'sq_entries': 32}, 1.045939, 2.1558, 2.25487, 2.35846, 35.5298),
    ('iq8', {'iq_entries': 8}, 1.428222, 1.8204, 2.59991, 3.71325, 35.5383),
    ('iq128', {'iq_entries': 128}, 0.8472, 2.5515, 2.16161, 1.83131, 35.6641),
    ('iq64_ancho2', {'iq_entries': 64, 'width': 2}, 0.974708, 2.2387, 2.18204, 2.12686, 35.527),
    ('iq64_lqsq32', {'iq_entries': 64, 'lq_entries': 32, 'sq_entries': 32}, 0.858842, 2.4711, 2.12229, 1.82271, 35.5253),
    ('iq64_rob256', {'iq_entries': 64, 'rob_entries': 256}, 0.846659, 2.6232, 2.22096, 1.88039, 36.5242),
    ('iq64_rob64', {'iq_entries': 64, 'rob_entries': 64}, 0.862933, 2.4404, 2.10588, 1.81724, 35.164),
    ('f2_l1i16', {'iq_entries': 64, 'l1i_kb': 16}, 0.849052, 2.5054, 2.12725, 1.80615, 35.438),
    ('f2_l1i32', {'iq_entries': 64, 'l1i_kb': 32}, 0.846995, 2.5142, 2.12949, 1.80366, 35.4615),
    ('f2_l1i128', {'iq_entries': 64, 'l1i_kb': 128}, 0.844739, 2.5348, 2.14125, 1.8088, 35.621),
    ('f2_l1d16', {'iq_entries': 64, 'l1d_kb': 16}, 0.850667, 2.4952, 2.12261, 1.80564, 35.4282),
    ('f2_l1d32', {'iq_entries': 64, 'l1d_kb': 32}, 0.848011, 2.5055, 2.12469, 1.80176, 35.4669),
    ('f2_l1d128', {'iq_entries': 64, 'l1d_kb': 128}, 0.846237, 2.597, 2.19768, 1.85976, 35.8893),
    ('f2_l2_256k', {'iq_entries': 64, 'l2_kb': 256}, 0.847006, 2.5208, 2.13517, 1.80851, 35.5415),
    ('f2_l2_512k', {'iq_entries': 64, 'l2_kb': 512}, 0.845189, 2.5447, 2.15073, 1.81777, 35.9101),
    ('f2_l2_2m', {'iq_entries': 64, 'l2_kb': 2048}, 0.84487, 2.662, 2.24903, 1.90014, 38.3152),
    ('f2_l2_4m', {'iq_entries': 64, 'l2_kb': 4096}, 0.844916, 2.8198, 2.38253, 2.01304, 42.6),
    ('f2_ancho3', {'iq_entries': 64, 'width': 3}, 0.957073, 2.2841, 2.18609, 2.09224, 35.5334),
    ('f2_ancho5', {'iq_entries': 64, 'width': 5}, 0.847138, 2.5447, 2.1557, 1.82617, 35.5505),
    ('f2_ancho6', {'iq_entries': 64, 'width': 6}, 0.846968, 2.5724, 2.17878, 1.84535, 35.5624),
    ('f2_iq48', {'iq_entries': 48}, 0.850502, 2.5046, 2.13015, 1.8117, 35.5403),
    ('f2_iq96', {'iq_entries': 96}, 0.847186, 2.5366, 2.149, 1.8206, 35.6012),
    ('f2_a5wb5', {'iq_entries': 64, 'width': 5, 'x_wb_width': 5}, 0.846881, 2.5456, 2.15585, 1.82575, 35.5505),
    ('f2_a6wb6', {'iq_entries': 64, 'width': 6, 'x_wb_width': 6}, 0.847414, 2.5729, 2.18033, 1.84764, 35.5624),
    ('f2_disp4', {'iq_entries': 64, 'x_dispatch_width': 4}, 0.847006, 2.5208, 2.13517, 1.80851, 35.5415),
    ('f2_issue2', {'iq_entries': 64, 'issue_width': 2}, 0.877459, 2.4227, 2.12585, 1.86535, 34.9447),
    ('f2_issue3', {'iq_entries': 64, 'issue_width': 3}, 0.850355, 2.4783, 2.10745, 1.79208, 35.0052),
    ('f2_issue4', {'iq_entries': 64, 'issue_width': 4}, 0.847317, 2.4933, 2.11264, 1.79008, 35.083),
    ('f2_issue6', {'iq_entries': 64, 'issue_width': 6}, 0.847151, 2.5073, 2.12409, 1.79943, 35.2797),
]

def cmd_import(store, args):
    nuevos = 0
    for name, ov, cpi, pw, en, edp, area in MANUAL:
        h = "m:" + phash(ov)
        if h in store.rows:
            print(f"  = {name}: misma configuración que '{store.rows[h]['name']}', se omite")
            continue
        store.add(make_row(ov, "manual", name, -1, len(store.rows) + 1, cpi, pw, en, edp, area))
        nuevos += 1
    print(f"Importadas {nuevos} simulaciones manuales. Total en {store.path.name}: {len(store.rows)}")

# ============================== CHECK ==============================
def cmd_check(args):
    """6 simulaciones reales. Las 3 primeras ya se midieron a mano: deben dar el mismo CPI."""
    args.root = str(SIMS.parent / "dse_check")
    path = RES / ("results_check_mock.csv" if args.mock else "results_check.csv")
    path.unlink(missing_ok=True)
    store = Store(path)
    plist = [S(), S(issue_width=3), S(issue_width=6), S(rob_entries=64), S(l1d_kb=32), S(iq_entries=48)]
    esperado = [0.847317, 0.850355, 0.847151]
    t0 = time.time()
    evaluate_batch(store, plist, "check", 0, args)
    print(f"\nTiempo total de las 6 en paralelo: {time.time() - t0:.0f} s")
    print(f"{'config':26s}{'CPI':>10s}{'esperado':>10s}  resultado")
    for i, p in enumerate(plist):
        r = store.rows.get(phash(p))
        if not r:
            print(f"{str(i):26s}  FALLÓ"); continue
        etq = f"issue={p['issue_width']} rob={p['rob_entries']} l1d={p['l1d_kb']} iq={p['iq_entries']}"
        if i < 3:
            ok = abs(float(r["cpi"]) - esperado[i]) < 6e-7
            print(f"{etq:26s}{float(r['cpi']):10.6f}{esperado[i]:10.6f}  {'IDÉNTICO' if ok else 'DIFERENTE'}")
        else:
            print(f"{etq:26s}{float(r['cpi']):10.6f}{'-':>10s}  nuevo (edp={float(r['edp']):.5f})")

# ============================== NSGA-II ==============================
def cmd_nsga(store, args):
    import numpy as np
    from pymoo.algorithms.moo.nsga2 import NSGA2
    from pymoo.core.evaluator import Evaluator
    from pymoo.core.problem import Problem
    from pymoo.operators.crossover.sbx import SBX
    from pymoo.operators.mutation.pm import PM
    from pymoo.operators.repair.rounding import RoundingRepair
    from pymoo.problems.static import StaticProblem

    n = len(KEYS)
    xu = np.array([len(SPACE[k]) - 1 for k in KEYS])
    problem = Problem(n_var=n, n_obj=2, xl=np.zeros(n), xu=xu.astype(float), vtype=int)
    rng = np.random.RandomState(args.seed)
    X0 = [encode(s) for s in SEEDS]
    while len(X0) < args.pop:
        X0.append(list(rng.randint(0, xu + 1)))
    algo = NSGA2(pop_size=args.pop, sampling=np.array(X0, dtype=float), eliminate_duplicates=True,
                 crossover=SBX(prob=0.9, eta=3, vtype=float, repair=RoundingRepair()),
                 mutation=PM(prob=0.2, eta=3, vtype=float, repair=RoundingRepair()))
    algo.setup(problem, termination=("n_gen", 10 ** 6), seed=args.seed, verbose=False)
    print(f"NSGA-II: población {args.pop}, semilla {args.seed}, objetivo {args.target} simulaciones nuevas "
          f"(ya hay {store.count('nsga')}), {args.workers} en paralelo", flush=True)
    gen = 0
    while True:
        pop = algo.ask()
        gen += 1
        plist = [decode(x) for x in pop.get("X")]
        antes = store.count("nsga")
        t0 = time.time()
        F = np.array(evaluate_batch(store, plist, "nsga", gen, args))
        Evaluator().eval(StaticProblem(problem, F=F), pop)
        algo.tell(infills=pop)
        b = store.best("edp")
        print(f"[gen {gen}] nuevas={store.count('nsga') - antes} únicas_nsga={store.count('nsga')}/{args.target} "
              f"mejor_EDP={float(b['edp']):.5f} ({time.time() - t0:.0f} s)", flush=True)
        if store.count("nsga") >= args.target:
            break
    print("NSGA-II terminado.")

# ============================ ALEATORIA ============================
def cmd_random(store, args):
    rng = random.Random(args.seed)
    vistos, orden = set(), []
    while len(orden) < args.n:
        idx = tuple(rng.randrange(len(SPACE[k])) for k in KEYS)
        if idx not in vistos:
            vistos.add(idx)
            orden.append({k: SPACE[k][i] for k, i in zip(KEYS, idx)})
    B = 30
    log = RES / ("random_log_mock.csv" if args.mock else "random_log.csv")
    for i in range(0, len(orden), B):
        evaluate_batch(store, orden[i:i + B], "random", i // B + 1, args)
        with open(log, "w", newline="") as f:      # orden de las propuestas aleatorias (para las curvas)
            w = csv.writer(f); w.writerow(["idx", "hash"])
            for j, p in enumerate(orden[:i + B]):
                w.writerow([j + 1, phash(p)])
        print(f"[aleatoria] {min(i + B, len(orden))}/{len(orden)}", flush=True)

# ============================== ESTADO ==============================
def cmd_status(store, args):
    print(f"Archivo: {store.path}  |  filas: {len(store.rows)}")
    for s in ("manual", "nsga", "random"):
        print(f"  {s:7s}: {store.count(s)}")
    for col in ("cpi", "energy", "edp"):
        b = store.best(col)
        if b:
            print(f"  mejor {col:6s}: {float(b[col]):.6f}  ({b['name'] or b['hash']}, fuente {b['source']})")
    ts = [r for r in store.rows.values() if r["source"] in ("nsga", "random") and r["gem5_wall_s"] not in ("", None)]
    if ts:
        g = sum(float(r["gem5_wall_s"]) for r in ts) / len(ts)
        m = sum(float(r["mcpat_s"]) for r in ts) / len(ts)
        print(f"  tiempo medio por simulación: gem5 {g:.0f} s, conversión+McPAT {m:.0f} s")

# ================================ MAIN ================================
def main():
    ap = argparse.ArgumentParser(description=__doc__, formatter_class=argparse.RawDescriptionHelpFormatter)
    ap.add_argument("cmd", choices=["import", "check", "nsga", "random", "status"])
    ap.add_argument("--workers", type=int, default=6)
    ap.add_argument("--mock", action="store_true", help="simulador falso (no usa gem5)")
    ap.add_argument("--target", type=int, default=1000, help="nsga: simulaciones nuevas a alcanzar")
    ap.add_argument("--pop", type=int, default=48, help="nsga: tamaño de población (múltiplo de 6 = rondas completas)")
    ap.add_argument("--seed", type=int, default=42)
    ap.add_argument("--n", type=int, default=300, help="random: número de configuraciones")
    args = ap.parse_args()
    args.root = str(SIMS)
    if args.cmd == "check":
        return cmd_check(args)
    store = Store(RES / ("results_mock.csv" if args.mock else "results.csv"))
    {"import": cmd_import, "nsga": cmd_nsga, "random": cmd_random, "status": cmd_status}[args.cmd](store, args)

if __name__ == "__main__":
    main()
````

### 9.5 NSGA-II frente a búsqueda aleatoria (mismo presupuesto: 300 evaluaciones)
**Comparación honesta.** Las dos búsquedas **no parten en igualdad**: el NSGA-II arranca con 6 semillas que salen de los experimentos manuales, y esos experimentos ya habían acotado el espacio.
Aun así, es la comparación relevante para el proyecto: *búsqueda guiada por conocimiento + evolución* frente a *muestreo uniforme del mismo espacio*.

````python
ref = (nm.cpi.max(), nm.energy.max())
N = 300
a, b = nsga.head(N), rnd.head(N)
tab = pd.DataFrame({
    "NSGA-II (300)": [a.edp.min(), a.cpi.min(), a.energy.min(), pareto_mask(a[["cpi","energy"]]).sum(), hypervolume(a[["cpi","energy"]], ref),
                      (a.edp <= nsga.edp.min()*1.01).sum(), a.edp.mean()],
    "Aleatoria (300)": [b.edp.min(), b.cpi.min(), b.energy.min(), pareto_mask(b[["cpi","energy"]]).sum(), hypervolume(b[["cpi","energy"]], ref),
                        (b.edp <= nsga.edp.min()*1.01).sum(), b.edp.mean()],
    "NSGA-II (1004)": [nsga.edp.min(), nsga.cpi.min(), nsga.energy.min(), pareto_mask(nsga[["cpi","energy"]]).sum(), hypervolume(nsga[["cpi","energy"]], ref),
                       (nsga.edp <= nsga.edp.min()*1.01).sum(), nsga.edp.mean()]},
    index=["Mejor EDP", "Mejor CPI", "Mejor energía", "Puntos en el frente", "Hipervolumen", "Configs a < 1 % del mejor EDP", "EDP medio"])
display(tab.round(5))

ns_n = [1] + list(range(10, len(nsga) + 1, 10)); rn_n = [1] + list(range(10, N + 1, 10))
fig, ax = plt.subplots(1, 3, figsize=(14, 4))
ax[0].plot(ns_n, [nsga.head(n).edp.min() for n in ns_n], color=AZUL, lw=2, label="NSGA-II")
ax[0].plot(rn_n, [rnd.head(n).edp.min() for n in rn_n], color=NARANJA, lw=2, label="Aleatoria")
ax[0].axhline(man.edp.min(), color=VERDE, ls="--", lw=1.2, label="Mejor manual")
ax[0].set_ylim(1.745, 1.80); ax[0].set_xlabel("Evaluaciones"); ax[0].set_ylabel("Mejor EDP hasta ahora"); ax[0].set_title("Convergencia del mejor EDP"); ax[0].legend()
ax[1].plot(ns_n, [hypervolume(nsga.head(n)[["cpi","energy"]], ref) for n in ns_n], color=AZUL, lw=2, label="NSGA-II")
ax[1].plot(rn_n, [hypervolume(rnd.head(n)[["cpi","energy"]], ref) for n in rn_n], color=NARANJA, lw=2, label="Aleatoria")
ax[1].set_xlabel("Evaluaciones"); ax[1].set_ylabel("Hipervolumen (mayor = mejor)"); ax[1].set_title("Calidad del frente"); ax[1].legend()
for src, col, nm_ in ((a, AZUL, "NSGA-II (300)"), (b, NARANJA, "Aleatoria (300)")):
    pf = src[pareto_mask(src[["cpi","energy"]])].sort_values("cpi")
    ax[2].step(pf.cpi, pf.energy, where="post", color=col, lw=2, label=nm_); ax[2].scatter(pf.cpi, pf.energy, s=18, color=col)
ax[2].set_xlim(0.843, 0.90); ax[2].set_ylim(2.02, 2.16); ax[2].set_xlabel("CPI"); ax[2].set_ylabel("Energía"); ax[2].set_title("Frentes con 300 evaluaciones"); ax[2].legend()
plt.tight_layout(); plt.savefig(FIG / "fig_convergencia.png", dpi=150); plt.show()
````

**Lectura (con la cautela que corresponde).**
- En el **mejor punto** la diferencia es pequeña: con 300 evaluaciones el mejor EDP aleatorio queda a ≈ 0.35 % del NSGA-II (≈ 0.5 % del mejor final). Obtener un EDP bueno es fácil en este espacio,
  porque la sensibilidad ya había acotado los rangos.
- En la **calidad del frente** sí gana el NSGA-II: más puntos no dominados (28 frente a 9), mayor hipervolumen y muchas más configuraciones a menos de 1 % del mejor EDP (41 frente a 3).
  La aleatoria desperdicia muestras en malos diseños (por ejemplo ancho 3).
- El NSGA-II **ya converge muy pronto**: con ~250 simulaciones alcanza ≈ 99 % del hipervolumen final. Las ~750 restantes mejoran el frente poco (≈ 1 %). Con un espacio tan acotado, 1000 simulaciones son más de las necesarias;
  se hicieron porque la guía exige al menos mil.

## 11. Velocidad de gem5 y McPAT
Por simulación (media, mediana, mín., máx.) y total. `gem5_wall_s` es el tiempo de reloj de gem5; `mcpat_s` incluye la conversión del profesor + McPAT.
El tiempo total se calcula como (última simulación terminada − primera) + duración media de una simulación, ya que las marcas de tiempo son de **fin** de simulación.

````python
rows = []
for nombre, df_ in (("NSGA-II", nsga), ("Aleatoria", d[d.source == "random"])):
    ts = pd.to_datetime(df_.timestamp)
    dur = (df_.gem5_wall_s + df_.mcpat_s)
    wall_h = ((ts.max() - ts.min()).total_seconds() + dur.mean()) / 3600
    rows.append({"Corrida": nombre, "Simulaciones": len(df_), "gem5 media (s)": df_.gem5_wall_s.mean(), "gem5 mediana (s)": df_.gem5_wall_s.median(),
                 "gem5 mín–máx (s)": f"{df_.gem5_wall_s.min():.0f}–{df_.gem5_wall_s.max():.0f}", "McPAT+conv. media (s)": df_.mcpat_s.mean(),
                 "McPAT % del total": 100 * df_.mcpat_s.sum() / dur.sum(), "Tiempo de CPU (h)": dur.sum() / 3600,
                 "Tiempo real (h)": wall_h, "Sims/hora": len(df_) / wall_h, "Aceleración": dur.sum() / 3600 / wall_h})
sp = pd.DataFrame(rows).set_index("Corrida"); display(sp.round(2).T)
fig, ax = plt.subplots(1, 2, figsize=(11, 3.4))
ax[0].hist(nsga.gem5_wall_s, bins=30, color=AZUL, alpha=.8); ax[0].set_xlabel("Tiempo de gem5 por simulación (s), 10 en paralelo"); ax[0].set_ylabel("Simulaciones")
ax[1].hist(nsga.mcpat_s, bins=30, color=NARANJA, alpha=.8); ax[1].set_xlabel("Conversión + McPAT por simulación (s)")
plt.tight_layout(); plt.savefig(FIG / "fig_velocidad.png", dpi=150); plt.show()
````

**Lectura.**
- Una simulación de gem5 con 10 en paralelo tarda ≈ 18 min (≈ 19 min con McPAT); **sola tarda ≈ 10 min** (593 s medidos en la fase 0) y con 6 a la vez ≈ 12 min. McPAT + conversión es ≈ 3 % del costo: **el tiempo lo manda gem5**.
- El Xeon tiene 6 núcleos físicos: 10 trabajadores rinden solo ≈ 8 % más que 6 (≈ 29 simulaciones/hora en ambos casos), porque los hilos extra ayudan poco y cada simulación va más lenta.
- Aceleración ≈ 9× sobre una ejecución en serie (el 100 % serían 10×). Perdida por esperar a la simulación más lenta de cada generación (≈ 8 %).
- Simulación individual en el PC de casa (fase 0): gem5 593 s, flujo completo 617 s (McPAT 21 s); en serie ≈ 5.8 simulaciones/hora; con 6 en paralelo ≈ 29.3/hora.

## 12. Lecciones aprendidas

> **BORRADOR — reescribe con tus propias palabras antes de entregar.** Está basado en lo que se midió, para que cada afirmación tenga respaldo.

1. **Entender antes de buscar.** Las 40 simulaciones manuales (≈ 3 % del trabajo total) fueron lo más valioso: mostraron que la IQ por defecto (16) era el cuello de botella y acotaron el espacio de 23 328 combinaciones.
   Sin eso, 1000 simulaciones habrían sido mucho menos informativas.
2. **El modelo de potencia manda lo que "conviene".** McPAT no ve los puertos de FU, el predictor, `dispatch_width` ni `wb_width`. Si se hubieran dejado libres, el optimizador habría elegido siempre el máximo "gratis".
   Hay que saber qué mide el modelo antes de fiarse del óptimo.
3. **Los parámetros interactúan.** El ancho 2 cuesta +4.5 % de CPI con IQ = 16 y +15 % con IQ = 64. Por eso variar de a uno no basta y una búsqueda multiobjetivo tiene sentido.
4. **Comparar con una línea base aleatoria da humildad.** El NSGA-II gana, pero en el mejor punto por solo ≈ 0.35 % con el mismo presupuesto; su ventaja real es el frente de Pareto (28 puntos frente a 9).
   Y converge con ~250 simulaciones: con un espacio ya acotado, buscar más rinde poco.
5. **Los resultados en el borde del espacio son una alarma.** SQ = 32 y L2 = 128 kB (los mínimos) aparecen en gran parte del frente: el óptimo de energía podría estar fuera del espacio explorado. Es la limitación principal.
6. **La reproducibilidad exige disciplina.** Una sola forma de lanzar gem5 (rutas absolutas + `--cwd`), caché por hash, y un entorno único para el DSE. Aun así, entre dos máquinas el CPI difiere ≈ 0.04 %.
7. **La velocidad la manda gem5, y los hilos extra no son núcleos.** McPAT es ≈ 3 % del tiempo; 10 trabajadores en 6 núcleos físicos dieron solo ≈ 8 % más que 6.
8. **Un solo workload.** `mp3_enc` tiene un conjunto de trabajo pequeño (por eso las cachés casi no importan). Las conclusiones sobre cachés pueden no valer para workloads más exigentes.

**Qué haría distinto:** incluir valores menores en el espacio (SQ 16, L2 64 kB), probar un segundo workload y recoger el perfil completo de la base en la misma máquina que el DSE.
